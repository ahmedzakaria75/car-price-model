# 2) Create Numerical Pipelines 


```python
num_cols = x.select_dtypes(include= 'number').columns
num_cols
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import RobustScaler

scaler = RobustScaler()

num_pipeline = Pipeline(steps= [ ('Scaling', scaler) ])
num_pipeline
```

### 3) Create Categorical Pipelines

```python
# check summary statistics for categorical columns
df.describe(include="object")
```

```python
x.select_dtypes(include= 'str').columns
```

### 3.1 OHE Pipeline

```python
ohe_cols = ['Gearbox']
ohe_cols
```

```python
from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(sparse_output= False, drop= 'first')

ohe_pipeline = Pipeline(steps= [ ('OHE', ohe) ])
ohe_pipeline
```

### 3.2 BE Pipeline

```python
be_cols = ['Location','Manufacturer Name','Car Type','Color','Energy']
be_cols
```

```python
from category_encoders import BinaryEncoder

be = BinaryEncoder()

be_pipeline = Pipeline(steps= [ ('BE', be) ])
be_pipeline
```

# 3.3 Ordinal Pipeline

```python
ord_cols = ['Feedback']
ord_cols
```

```python
from sklearn.preprocessing import OrdinalEncoder

ord = OrdinalEncoder(categories= [ ['Excellent', 'Good', 'Average', 'Poor'] ])

ord_pipeline = Pipeline(steps= [ ('Ord', ord) ])
ord_pipeline
```

# 4) Assign each column to corresponding Pipeline

```python
from sklearn.compose import ColumnTransformer

preprocessing = ColumnTransformer(transformers= [
    ('Num Pipeline', num_pipeline, num_cols),
    ('OHE Pipeline', ohe_pipeline, ohe_cols),
    ('BE Pipeline', be_pipeline, be_cols),
    ('Ordinal Pipeline', ord_pipeline, ord_cols)
], remainder= 'passthrough')                        

preprocessing
```

```python
from sklearn.linear_model import LinearRegression

le = LinearRegression()

model_pipeline = Pipeline(steps= [ ('Preprocessing', preprocessing), ('Model', le) ])
model_pipeline
```

```python
from sklearn.model_selection import cross_validate

cross_val = cross_validate(model_pipeline, x, y, scoring= 'r2', cv= 5, return_train_score= True)
cross_val
```

```python
print('Avg Train Score :', cross_val['train_score'].mean().round(4) * 100)

print('Avg Test Score :', cross_val['test_score'].mean().round(4) * 100)
```

```python
import joblib

joblib.dump(model_pipeline, 'car_price_model.joblib')
```

```python
from sklearn.linear_model import LinearRegression, SGDRegressor, Lasso, Ridge
from sklearn.neighbors import KNeighborsRegressor
from sklearn.svm import SVR
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import cross_validate

models = [
    ('Linear Regression', LinearRegression()),
    ('SGD Regressor', SGDRegressor(random_state= 42)),
    ('Lasso', Lasso(random_state= 42)),
    ('Ridge', Ridge(random_state= 42)),
    ('KNN Regressor', KNeighborsRegressor()),
    ('SVR', SVR()),
    ('Decision Tree Regressor', DecisionTreeRegressor(random_state= 42)),
    ('Random Forest Regressor', RandomForestRegressor(random_state= 42)),
]

for model in models:

    

    model_pipeline = Pipeline(steps= [ ('Preprocessing', preprocessing), ('Model', model[1]) ])

    cv_scores = cross_validate(model_pipeline, x, y, cv= 5, scoring= 'r2', return_train_score= True)

    print(model[0])
    print('Train Score: ', round(cv_scores['train_score'].mean() * 100, 3))
    print('Test Score: ', round(cv_scores['test_score'].mean() * 100, 3))
    print('\n', '-' * 100, '\n')
```
